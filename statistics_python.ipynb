# 1. 1.	The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance

```python
import math

# Given values
mu = 65          # Historical average
x_bar = 68       # Sample mean
s = 10           # Sample standard deviation
n = 50           # Sample size
alpha = 0.05     # Significance level

# Calculate Z-score
z = (x_bar - mu) / (s / math.sqrt(n))

print("Z-score:", z)

# Critical value for 5% right-tailed test
critical_value = 1.645

print("Critical value:", critical_value)

# Decision
if z > critical_value:
    print("Reject H0")
    print("The new course improves performance.")
else:
    print("Fail to reject H0")
    print("There is not enough evidence that the new course improves performance.")
```

# 3.	A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed

```python
import math

# Given values
p0 = 0.60
x = 134
n = 200
alpha = 0.05

# Step 1: Calculate sample proportion
p_hat = x / n

print("Sample proportion:", p_hat)

# Step 2: Calculate Z-score
z = (p_hat - p0) / math.sqrt((p0 * (1 - p0)) / n)

print("Z-score:", z)

# Step 3: Critical value
critical_value = 1.645

print("Critical value:", critical_value)

# Step 4: Decision
if z > critical_value:
    print("Reject H0")
    print("Customer satisfaction is significantly higher than 60%.")
else:
    print("Fail to reject H0")
    print("There is not enough evidence that satisfaction is higher than 60%.")
```

# 5.	Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

```python
from scipy import stats

# Method A
n1 = 30
mean1 = 75
sd1 = 12

# Method B
n2 = 28
mean2 = 70
sd2 = 10

# Significance level
alpha = 0.05

# Two-sample t-test
t_stat, p_value = stats.ttest_ind_from_stats(
    mean1, sd1, n1,
    mean2, sd2, n2,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject H0")
    print("There is a significant difference between the two methods.")
else:
    print("Fail to reject H0")
    print("There is no significant difference between the two methods.")
```

# 6.	The average delivery time from two warehouses is compared using random samples. Warehouse A has a sample of 35 deliveries with a mean time of 48 minutes and a standard deviation of 6 minutes. Warehouse B has a sample of 40 deliveries with a mean time of 52 minutes and a standard deviation of 7 minutes. At the 5% level of significance, test whether the mean delivery times differ.

```python
from scipy import stats

# Given values

# Warehouse A
n1 = 35
mean1 = 48
sd1 = 6

# Warehouse B
n2 = 40
mean2 = 52
sd2 = 7

# Significance level
alpha = 0.05

# Perform independent two-sample t-test
t_stat, p_value = stats.ttest_ind_from_stats(
    mean1, sd1, n1,
    mean2, sd2, n2,
    equal_var=False
)

# Display results
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject H0")
    print("There is a significant difference in mean delivery times.")
else:
    print("Fail to reject H0")
    print("There is no significant difference in mean delivery times.")
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Given values

# Version A
clicks_A = 320
users_A = 500

# Version B
clicks_B = 290
users_B = 480

# Significance level
alpha = 0.05

# Calculate conversion rates
rate_A = clicks_A / users_A
rate_B = clicks_B / users_B

print("Version A conversion rate:", rate_A)
print("Version B conversion rate:", rate_B)

# Two-proportion z-test
z_stat, p_value = proportions_ztest(
    [clicks_A, clicks_B],
    [users_A, users_B],
    alternative='two-sided'
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject H0")
    print("There is a significant difference in conversion rates.")
else:
    print("Fail to reject H0")
    print("There is no significant difference in conversion rates.")
```

# 8.A vaccine is tested in two regions. In Region A, 180 out of 300 people recover. In Region B, 165 out of 310 people recover. At the 5% level of significance, test whether the recovery rates are significantly different.


```python
import math

# Region A
recover_A = 180
total_A = 300

# Region B
recover_B = 165
total_B = 310

# Significance level
alpha = 0.05

# Step 1: Calculate recovery rates
p1 = recover_A / total_A
p2 = recover_B / total_B

print("Region A recovery rate:", p1)
print("Region B recovery rate:", p2)

# Step 2: Calculate pooled proportion
p_pool = (recover_A + recover_B) / (total_A + total_B)

print("Pooled proportion:", p_pool)

# Step 3: Calculate standard error
se = math.sqrt(
    p_pool * (1 - p_pool) *
    (1/total_A + 1/total_B)
)

print("Standard error:", se)

# Step 4: Calculate Z-statistic
z = (p1 - p2) / se

print("Z-statistic:", z)

# Step 5: Critical value
critical_value = 1.96

print("Critical value:", critical_value)

# Step 6: Decision
if abs(z) > critical_value:
    print("Reject H0")
    print("The recovery rates are significantly different.")
else:
    print("Fail to reject H0")
    print("The recovery rates are not significantly different.")
```


# 9.	A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.


```python
import math
from scipy.stats import f

# Program A
n_A = 25
mean_A = 80
sd_A = 8

# Program B
n_B = 30
mean_B = 75
sd_B = 7

# Program C
n_C = 28
mean_C = 82
sd_C = 6

# Significance level
alpha = 0.05

# Total number of employees
total_n = n_A + n_B + n_C

# Grand Mean
grand_mean = (
    n_A * mean_A +
    n_B * mean_B +
    n_C * mean_C
) / total_n

print("Grand Mean:", grand_mean)

# Between-group variation
SS_between = (
    n_A * (mean_A - grand_mean)**2 +
    n_B * (mean_B - grand_mean)**2 +
    n_C * (mean_C - grand_mean)**2
)

print("SS Between:", SS_between)

# Within-group variation
SS_within = (
    (n_A - 1) * sd_A**2 +
    (n_B - 1) * sd_B**2 +
    (n_C - 1) * sd_C**2
)

print("SS Within:", SS_within)

# Degrees of freedom
df_between = 3 - 1
df_within = total_n - 3

print("DF Between:", df_between)
print("DF Within:", df_within)

# Mean squares
MS_between = SS_between / df_between
MS_within = SS_within / df_within

print("MS Between:", MS_between)
print("MS Within:", MS_within)

# F-statistic
F_stat = MS_between / MS_within

print("F-statistic:", F_stat)

# Critical F-value
F_critical = f.ppf(1 - alpha, df_between, df_within)

print("F-critical:", F_critical)

# Decision
if F_stat > F_critical:
    print("Reject H0")
    print("There is a significant difference in mean productivity across the programs.")
else:
    print("Fail to reject H0")
    print("There is no significant difference in mean productivity across the programs.")
```

# 10.	A company wants to test whether gender and product preference are independent. The survey results are as follows:
1.	Male: 60 prefer Product A, 40 prefer Product B
2.	Female: 50 prefer Product A, 50 prefer Product B
At the 5% level of significance, test for independence.


```python
import numpy as np
from scipy.stats import chi2_contingency

# Observed data
data = np.array([
    [60, 40],   # Male
    [50, 50]    # Female
])

# Significance level
alpha = 0.05

# Chi-square test
chi2, p_value, dof, expected = chi2_contingency(data)

# Display results
print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)

print("Expected values:")
print(expected)

# Decision
if p_value < alpha:
    print("Reject H0")
    print("Gender and product preference are not independent.")
else:
    print("Fail to reject H0")
    print("Gender and product preference are independent.")
```

# 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
1.	Face 1: 18
2.	Face 2: 22
3.	Face 3: 20
4.	Face 4: 25
5.	Face 5: 17
6.	Face 6: 18
At the 5% level of significance, test whether the die is fair.


```python
import numpy as np
from scipy.stats import chisquare

# Observed frequencies
observed = np.array([18, 22, 20, 25, 17, 18])

# Expected frequencies
expected = np.array([20, 20, 20, 20, 20, 20])

# Significance level
alpha = 0.05

# Chi-square goodness-of-fit test
chi2, p_value = chisquare(
    observed,
    f_exp=expected
)

# Display results
print("Chi-square statistic:", chi2)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject H0")
    print("The die is not fair.")
else:
    print("Fail to reject H0")
    print("The die can be considered fair.")
```
